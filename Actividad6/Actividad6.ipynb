```python
df2["age"] =df2["age"].fillna(0) 
df2["embark_town"] =df2["embark_town"].fillna('unknown') 
df2["embarked"] =df2["embarked"].fillna('unknown') 
df2["deck"] =df2["deck"].fillna('unknown') 
valores_nulos=df2.isnull().sum().sum()
valores_nulos
```

```python
df3= pd.read_csv('Step_2.csv')
df3.info()
```

```python
valores_nulos=df3.isnull().sum().sum()
valores_nulos
```

```python
cuantitativas = df3.select_dtypes(include=['float64', 'int64'])
cualitativas = df3.select_dtypes(exclude=['float64', 'int64'])
```

```python
fig = plt.figure(figsize=(15, 8))
cuantitativas.plot(kind='box', vert=False, figsize=(15, 8))
plt.title("Valores Atípicos de STEP_2 (Antes de la sustitución)")
plt.show()
```

```python
y = cuantitativas

percentile25 = y.quantile(0.25) 
percentile75 = y.quantile(0.75) 
iqr = percentile75 - percentile25

Limite_Superior_iqr = percentile75 + 1.5 * iqr
Limite_Inferior_iqr = percentile25 - 1.5 * iqr

data3_iqr = cuantitativas[(y <= Limite_Superior_iqr) & (y >= Limite_Inferior_iqr)]

data4_iqr = data3_iqr.copy()
data4_iqr = data4_iqr.fillna(round(data3_iqr.median(), 1))

Datos_limpios = pd.concat([cualitativas, data4_iqr], axis=1)

valores_nulos = Datos_limpios.isnull().sum()
valores_nulos
```

```python
Datos_limpios.to_csv("STEP_3.csv", index=False)
```
